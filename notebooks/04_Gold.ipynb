```python
from src.config import OUTPUT_FILE

df = pd.read_csv(OUTPUT_FILE)

df.head()
```

```python
import numpy as np

## INDICADORES FINANCIEROS

# Crecimiento de ingresos (%)
df["crecimiento_ingresos_pct"] = np.where(
    df["ingresos_operacionales_2024"] >= 1000,
    (
        (df["ingresos_operacionales_2025"] - df["ingresos_operacionales_2024"])
        / df["ingresos_operacionales_2024"]
    ) * 100,
    0
)

# Margen de utilidad (%)
df["margen_utilidad_pct"] = np.where(
    df["ingresos_operacionales_2025"] >= 1000,
    (
        df["ganancia_pérdida_2025"]
        / df["ingresos_operacionales_2025"]
    ) * 100,
    0
)

# Endeudamiento (%)
df["endeudamiento_pct"] = np.where(
    df["total_activos_2025"] >= 1000,
    (
        df["total_pasivos_2025"]
        / df["total_activos_2025"]
    ) * 100,
    0
)

# ROA (%)
df["roa_pct"] = np.where(
    df["total_activos_2025"] >= 1000,
    (
        df["ganancia_pérdida_2025"]
        / df["total_activos_2025"]
    ) * 100,
    0
)

# ROE (%)
df["roe_pct"] = np.where(
    df["total_patrimonio_2025"].abs() >= 1000,
    (
        df["ganancia_pérdida_2025"]
        / df["total_patrimonio_2025"]
    ) * 100,
    0
)

# LIMPIEZA FINAL

# Reemplazar infinitos por NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
gold = df[
    [
        "ranking_2025",
        "nit",
        "razon_social",
        "región",
        "departamento_domicilio",
        "ciudad_domicilio",
        "macrosector",
        "ingresos_operacionales_2025",
        "ganancia_pérdida_2025",
        "total_activos_2025",
        "total_pasivos_2025",
        "total_patrimonio_2025",
        "crecimiento_ingresos_pct",
        "margen_utilidad_pct",
        "endeudamiento_pct",
        "roa_pct",
        "roe_pct"
    ]
]

gold.head()
```
