```python
import pandas as pd

gold = pd.read_csv(
    r"C:\Users\USUARIO\Documents\GitHub\Empresas-Colombianas\data\Gold\empresas_colombia_gold_Final.csv"
)

print(gold.columns)
```

```python
import pandas as pd
import numpy as np

# ==========================================================
# 1. INFORMACIÓN GENERAL
# ==========================================================

print("\n1. INFORMACIÓN GENERAL")

print(f"Filas: {gold.shape[0]}")
print(f"Columnas: {gold.shape[1]}")

# ==========================================================
# 2. TIPOS DE DATOS
# ==========================================================

print("\n2. TIPOS DE DATOS")

print(gold.dtypes)

# ==========================================================
# 3. VALORES NULOS
# ==========================================================

print("\n3. VALORES NULOS")

nulos = gold.isnull().sum()

print(nulos)

# ==========================================================
# 4. VALORES DUPLICADOS
# ==========================================================

print("\n4. REGISTROS DUPLICADOS")

duplicados = gold.duplicated().sum()

print(f"Duplicados: {duplicados}")

# ==========================================================
# 5. VALORES INFINITOS
# ==========================================================

print("\n5. VALORES INFINITOS")

columnas_numericas = gold.select_dtypes(include=np.number).columns

for columna in columnas_numericas:

    cantidad = np.isinf(gold[columna]).sum()

    if cantidad > 0:

        print(f"{columna}: {cantidad}")

# ==========================================================
# 6. ESTADÍSTICAS GENERALES
# ==========================================================

print("\n6. ESTADÍSTICAS DESCRIPTIVAS")

print(gold.describe().T)

# ==========================================================
# 7. AUDITORÍA DE INDICADORES
# ==========================================================

indicadores = [

    "crecimiento_ingresos_pct",
    "margen_utilidad_pct",
    "endeudamiento_pct",
    "roa_pct",
    "roe_pct"

]

print("\n7. AUDITORÍA DE INDICADORES")

for indicador in indicadores:

    print("\n" + "="*70)

    print(indicador)

    print("="*70)

    print(f"Valor mínimo : {gold[indicador].min():,.4f}")

    print(f"Valor máximo : {gold[indicador].max():,.4f}")

    print(f"Promedio     : {gold[indicador].mean():,.4f}")

    print(f"Mediana      : {gold[indicador].median():,.4f}")

# ==========================================================
# 8. TOP 10 VALORES MÁS ALTOS
# ==========================================================

print("\n8. TOP 10 VALORES MÁS ALTOS")

for indicador in indicadores:

    print("\n" + "="*80)

    print(f"TOP 10 - {indicador}")

    print("="*80)

    print(

        gold.nlargest(10, indicador)[

            [

                "ranking_2025",
                "nit",
                "razon_social",

                indicador

            ]

        ]

    )

# ==========================================================
# 9. TOP 10 VALORES MÁS BAJOS
# ==========================================================

print("\n9. TOP 10 VALORES MÁS BAJOS")

for indicador in indicadores:

    print("\n" + "="*80)

    print(f"BOTTOM 10 - {indicador}")

    print("="*80)

    print(

        gold.nsmallest(10, indicador)[

            [

                "ranking_2025",
                "nit",
                "razon_social",

                indicador

            ]

        ]

    )

# ==========================================================
# 10. REVISIÓN DE OUTLIERS (IQR)
# ==========================================================

print("\n10. OUTLIERS")

for indicador in indicadores:

    q1 = gold[indicador].quantile(.25)

    q3 = gold[indicador].quantile(.75)

    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr

    limite_superior = q3 + 1.5 * iqr

    outliers = gold[
        (gold[indicador] < limite_inferior)
        |
        (gold[indicador] > limite_superior)
    ]

    print(f"\n{indicador}")

    print(f"Cantidad de outliers: {len(outliers)}")

# ==========================================================
# 11. VALIDACIÓN FINAL
# ==========================================================

print("\n11. VALIDACIÓN FINAL")

errores = False

if gold.isnull().sum().sum() > 0:

    print("Existen valores nulos")

    errores = True

if gold.duplicated().sum() > 0:

    print("Existen registros duplicados")

    errores = True

if np.isinf(gold.select_dtypes(include=np.number)).sum().sum() > 0:

    print("Existen valores infinitos")

    errores = True

if errores:

    print("\nREVISAR EL DATASET ANTES DE CARGARLO AL DATA WAREHOUSE")

else:

    print("\nDATASET APROBADO PARA CARGA AL DATA WAREHOUSE")
```
